```sql
CREATE OR REPLACE VIEW gold.petrol_prices.latest_prices_metric_view
WITH METRICS
LANGUAGE YAML
AS $$
  version: 1.1

  source: gold.petrol_prices.latest_prices

  joins:
    - name: postcodes_dim
      source: gold.petrol_prices.postcodes
      "on": source.postcode = postcodes_dim.postcode
    - name: fuel_types_dim
      source: gold.petrol_prices.fuel_types
      "on": source.fuel_type_code = fuel_types_dim.fuel_type_code
    - name: forecourts_dim
      source: gold.petrol_prices.active_forecourts
      "on": source.forecourt_id = forecourts_dim.forecourt_id

  dimensions:
    - name: forecourt_id
      expr: source.forecourt_id
      display_name: Forecourt Id
    - name: trading_name
      expr: forecourts_dim.trading_name
      display_name: Trading Name
    - name: brand_name
      expr: forecourts_dim.brand_name
      display_name: Brand Name
    - name: forecourt_type
      expr: forecourts_dim.forecourt_type
      display_name: Forecourt Type
    - name: postcode
      expr: forecourts_dim.postcode
      display_name: Postcode
    - name: fuel_type_description
      expr: fuel_types_dim.fuel_type_description
      display_name: Fuel Type
    - name: price
      expr: source.price
      display_name: Price (pence per litre)
    - name: geographic_postcode
      expr: postcodes_dim.geographic_postcode
    - name: outward_postcode
      expr: postcodes_dim.outward_postcode
    - name: latitude
      expr: postcodes_dim.latitude
    - name: longitude
      expr: postcodes_dim.longitude

  measures:
    - name: count
      expr: COUNT(*)
      display_name: Count
    - name: median_price
      expr: MEDIAN(source.price)
      display_name: Median Price
$$
```

```sql
CREATE OR REPLACE VIEW gold.petrol_prices.price_history_metric_view
WITH METRICS
LANGUAGE YAML
AS $$
  version: 1.1

  source: gold.petrol_prices.price_history

  joins:
    - name: postcodes
      source: gold.petrol_prices.postcodes
      "on": source.postcode = postcodes.postcode
    - name: dates
      source: gold.petrol_prices.dates
      "on": source.date_key = dates.date_key
    - name: forecourt_history
      source: gold.petrol_prices.forecourt_history
      "on": source.forecourt_entry_id = forecourt_history.entry_id
    - name: fuel_types
      source: gold.petrol_prices.fuel_types
      "on": source.fuel_type_code = fuel_types.fuel_type_code

  dimensions:
    - name: last_update
      expr: source.last_update
      display_name: Last Update
    - name: date_key
      expr: source.date_key
      display_name: Date Key
    - name: postcode
      expr: source.postcode
      display_name: Postcode
    - name: forecourt_id
      expr: source.forecourt_id
      display_name: Forecourt Id
    - name: fuel_type_code
      expr: source.fuel_type_code
      display_name: Fuel Type Code
    - name: price
      expr: source.price
      display_name: Price
    - name: date
      expr: dates.date
      display_name: Price Date
    - name: geographic_postcode
      expr: postcodes.geographic_postcode
      display_name: Geographic Postcode
    - name: outward_postcode
      expr: postcodes.outward_postcode
      display_name: Outward Postcode
    - name: trading_name
      expr: forecourt_history.trading_name
      display_name: Trading Name
    - name: brand_name
      expr: forecourt_history.brand_name
      display_name: Brand Name
    - name: forecourt_type
      expr: forecourt_history.forecourt_type
      display_name: Forecourt Type
    - name: fuel_type_description
      expr: fuel_types.fuel_type_description
      display_name: Fuel Type

  measures:
    - name: count
      expr: COUNT(*)
      display_name: Fuel Station Count
    - name: mean
      expr: AVG(source.price)
      display_name: Mean fuel price
    - name: lower_quartile
      expr: "PERCENTILE(source.price, 0.25)"
      display_name: Lower quartile of fuel price
    - name: upper_quartile
      expr: "PERCENTILE(source.price, 0.75)"
      display_name: Upper quartile of fuel price
    - name: max
      expr: MAX(source.price)
      display_name: Highest fuel price
    - name: min
      expr: MIN(source.price)
      display_name: Lowest fuel price
    - name: stddev
      expr: STDDEV(source.price)
      display_name: Standard Deviation of fuel prices
$$
```
